## DuckDB Raw Upload

```python
-- SQL instructions to import boxoffice raw data

CREATE SCHEMA IF NOT EXISTS raw;

DROP TABLE IF EXISTS raw.boxoffice_movies;

CREATE TABLE raw.boxoffice_movies AS
SELECT *
-- read_csv_auto is a duckdb function to make the CSV file queryable like a table
FROM read_csv_auto(
    '/home/alessio/Documents/ELT_GAV/data/raw/moviesboxoffice.csv',
    -- tells duckdb that first row contain the column names
    header = true
);
```
